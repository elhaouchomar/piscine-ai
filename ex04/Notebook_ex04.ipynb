```python
import spacy

nlp = spacy.load("en_core_web_sm")

s1 = nlp("I want to buy shoes")
s2 = nlp("I would love to purchase running shoes")
s3 = nlp("I am in my room")

print(f"sentence_1 <=> sentence 2 : {s1.similarity(s2)}")
print(f"sentence_1 <=> sentence 3 : {s1.similarity(s3)}")
print(f"sentence_2 <=> sentence 3 : {s2.similarity(s3)}")
```
