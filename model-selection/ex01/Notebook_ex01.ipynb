```python
import numpy as np
from sklearn.model_selection import KFold

X = np.array(np.arange(1,21).reshape(10,-1))
y = np.array(np.arange(1,11))

kf = KFold(n_splits=5)

fold = 1
for train_index, test_index in kf.split(X):
    print(f"Fold:  {fold}")
    print("TRAIN:", train_index, "TEST:", test_index)
    print()
    fold += 1
```
