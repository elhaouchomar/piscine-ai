```python
import pandas as pd
import numpy as np
import plotly.express as px


returns = np.random.randn(50)
price = 100 + np.cumsum(returns)
dates = pd.date_range(start='2020-09-01', periods=50, freq='B')
df = pd.DataFrame({'Date': dates, 'Company_A': price})

fig_px = px.line(df, x='Date', y='Company_A', title="Company_1 stock price",
                 labels={'Date':'Date','Company_A':'Price'})
fig_px.show()
```
