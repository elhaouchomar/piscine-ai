```python
import numpy as np

class Neuron:
    def __init__(self, w1, w2, b):
        self.w1 = w1
        self.w2 = w2
        self.b = b

    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))

    def feedforward(self, x1, x2):
        return self.sigmoid(x1 * self.w1 + x2 * self.w2 + self.b)


def log_loss_custom(y_true, y_pred):
    y_pred = np.clip(y_pred, 1e-15, 1 - 1e-15)
    return -np.mean(
        y_true * np.log(y_pred) +
        (1 - y_true) * np.log(1 - y_pred)
    )


if __name__ == "__main__":

    neuron_h1 = Neuron(0.05, 0.001, 0)
    neuron_h2 = Neuron(0.02, 0.003, 0)
    neuron_o1 = Neuron(2, 0, 0)

    X = np.array([
        [12, 15],
        [10, 9],
        [18, 18],
        [13, 14]
    ])

    y_true = np.array([1, 0, 1, 1])

    outputs = []

    for x1, x2 in X:
        h1 = neuron_h1.feedforward(x1, x2)
        h2 = neuron_h2.feedforward(x1, x2)
        y = neuron_o1.feedforward(h1, h2)
        outputs.append(y)

    outputs = np.array(outputs)

    names = ["Bob", "Eli", "Tom", "Ryan"]

    for name, value in zip(names, outputs):
        print(f"{name}: {value}")

    print("Log loss:", log_loss_custom(y_true, outputs))
```
