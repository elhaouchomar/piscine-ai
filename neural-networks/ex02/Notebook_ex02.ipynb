```python
import numpy as np

class Neuron:
    def __init__(self, w1, w2, b):
        self.w1 = w1
        self.w2 = w2
        self.b = b

    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))

    def feedforward(self, x1, x2):
        return self.sigmoid(x1 * self.w1 + x2 * self.w2 + self.b)


class OurNeuralNetwork:
    def __init__(self, h1, h2, o1):
        self.h1 = h1
        self.h2 = h2
        self.o1 = o1

    def feedforward(self, x1, x2):
        h1_out = self.h1.feedforward(x1, x2)
        h2_out = self.h2.feedforward(x1, x2)
        return self.o1.feedforward(h1_out, h2_out)


if __name__ == "__main__":
    neuron_h1 = Neuron(1, 2, -1)
    neuron_h2 = Neuron(0.5, 1, 0)
    neuron_o1 = Neuron(2, 0, 1)

    nn = OurNeuralNetwork(neuron_h1, neuron_h2, neuron_o1)
    print(nn.feedforward(2, 3))
```
