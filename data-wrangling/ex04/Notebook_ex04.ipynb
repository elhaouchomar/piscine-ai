```python
import pandas as pd
import numpy as np

def winsorize(df, quantiles):
    """
    df: pd.DataFrame
    quantiles: list ex: [0.20, 0.80]
    """
    lower_bound = np.quantile(df, quantiles[0])
    upper_bound = np.quantile(df, quantiles[1])
    
    return df.clip(lower=lower_bound, upper=upper_bound)

df_test = pd.DataFrame(range(1,11), columns=['sequence'])
print(winsorize(df_test, [0.20, 0.80]).to_markdown())
```

```python
groups = np.concatenate([np.ones(10), np.ones(10)+1, np.ones(10)+2, np.ones(10)+3, np.ones(10)+4])
df = pd.DataFrame(data=zip(groups, range(1, 51)), columns=["group", "sequence"])

output = df.groupby("group", group_keys=False)[['sequence']].apply(winsorize, quantiles=[0.05, 0.95])

print(output.head(11).to_markdown())
```
