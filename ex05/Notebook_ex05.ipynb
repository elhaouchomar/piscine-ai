```python
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
X, y, coef = make_regression(n_samples=100,
                         n_features=1,
                         n_informative=1,
                         noise=10,
                         coef=True,
                         random_state=0,
                         bias=100.0
)

plt.scatter(X, y)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('x (1 dimension) and y')
plt.show()
```
