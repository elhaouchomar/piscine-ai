```python
import numpy as np

class Neuron:
    def __init__(self, w1, w2, b, regression=False):
        self.w1 = w1
        self.w2 = w2
        self.b = b
        self.regression = regression

    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))

    def activate(self, x):
        if self.regression:
            return x
        return self.sigmoid(x)

    def feedforward(self, x1, x2):
        z = x1 * self.w1 + x2 * self.w2 + self.b
        return self.activate(z)


if __name__ == "__main__":

    neuron = Neuron(0, 1, 4, True)
    print(neuron.feedforward(2, 3))

    X = np.array([
        [12, 15],
        [10, 9],
        [18, 18],
        [13, 14]
    ])

    y_true = np.array([16, 10, 19, 16])

    neuron_h1 = Neuron(0.05, 0.001, 0, False)
    neuron_h2 = Neuron(0.002, 0.003, 0, False)
    neuron_o1 = Neuron(2, 7, 10, True)

    outputs = []

    for x1, x2 in X:
        h1 = neuron_h1.feedforward(x1, x2)
        h2 = neuron_h2.feedforward(x1, x2)
        y = neuron_o1.feedforward(h1, h2)
        outputs.append(y)

    outputs = np.array(outputs)

    names = ["Bob", "Eli", "Tom", "Ryan"]

    for name, value in zip(names, outputs):
        print(f"{name}: {value}")

    mse = np.mean((outputs - y_true) ** 2)
    print(mse)
```
