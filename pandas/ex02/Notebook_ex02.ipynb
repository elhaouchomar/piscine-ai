```python
import pandas as pd

df = pd.read_csv("data/household_power_consumption.txt", sep=';', na_values='?')
df.drop(["Time","Sub_metering_2","Sub_metering_3"], axis=1, inplace=True)

#  not a solution I recommend becquse drop one by one
# del df["Time"]
# del df["Sub_metering_2"]
# del df["Sub_metering_3"]
```

```python
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)
df.set_index('Date', inplace=True)

# test
df.head().index
```

```python
def update_types(df):
    for col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')
    return df

df = update_types(df)

#test
df.dtypes
```

```python
df.dropna(inplace=True)
```
