```python
from sklearn.linear_model import LogisticRegression
import numpy as np
X = [[0],[0.1],[0.2], [1],[1.1],[1.2], [1.3]]
y = [0,0,0,1,1,1,0]

x_pred = [[0.5]]

model = LogisticRegression()
model.fit(X, y)

print("Predict the class for x_pred = [[0.5]]:", model.predict(x_pred))
print("Predict the probabilities for x_pred = [[0.5]]:", model.predict_proba(x_pred))
print("rint the coefficients (coef_):", model.coef_(x_pred))
```
