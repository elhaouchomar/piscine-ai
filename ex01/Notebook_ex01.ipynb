```python
import numpy as np

class Neuron:
    def __init__(self, weight1, weight2, bias):
        self.w1 = weight1
        self.w2 = weight2
        self.b = bias

    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))

    def feedforward(self, x1, x2):
        z = x1 * self.w1 + x2 * self.w2 + self.b
        return self.sigmoid(z)


if __name__ == "__main__":
    neuron = Neuron(0, 1, 4)
    print(neuron.feedforward(2, 3))
```
