```python
import numpy as np

arr = np.ones((9, 9), dtype=np.int8)
print(arr)
print(arr.dtype)

print("=" * 72)

# Step 1: Create 9x9 array of ones
arr = np.ones((9, 9), dtype=np.int8)

# Step 2: Make the inner area zero
arr[1:-1, 1:-1] = 0

# Step 3: Create the inner square of ones
arr[2:-2, 2:-2] = 1

# Step 4: Create the smaller inner zero square
arr[3:-3, 3:-3] = 0

# Step 5: Center element = 1
arr[4, 4] = 1

print(arr)
```
