```python
import itertools
import numpy as np

model = np.genfromtxt("./data/model_forecasts.txt", delimiter=",", dtype=np.float32)
model = np.nan_to_num(model, nan=0.0)
print(model)

print("=" * 72)
teams = np.arange(10)
perms = np.array(list(itertools.permutations(teams)))
matches = perms.reshape(-1, 5, 2)
print(matches)


print("=" * 72)
scores = model[matches[:, :, 0], matches[:, :, 1]]
print(scores)

print("=" * 72)
total_scores = np.sum(scores ** 2, axis=1)
best_idx = np.argmin(total_scores)
best_matches = matches[best_idx]
res = best_matches.T
print(res)
```
