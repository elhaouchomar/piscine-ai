```python
import numpy as np

data = np.genfromtxt("./data/winequality-red.csv", 
                     delimiter=";",
                     skip_header=1,
                     dtype=np.float32
                    )

print("1 " + "=" * 72)
print(data.nbytes)
print(data)


print("\n2 " + "=" * 72)
rows = data[[1, 6, 11]]
rows = rows[~np.isnan(rows)]
print(rows)


print("\n3 " + "=" * 72)
alco = data[:,-1]
res = np.any(alco > 20)
print(res)


print("\n4 " + "=" * 72)
mean_alco = np.mean(data[:, -1])
print(mean_alco)


print("\n5 " + "=" * 72)
ph = data[:, 8]
stats = {
    "min": np.nanmin(ph),
    "max": np.nanmax(ph),
    "mean": np.nanmean(ph),
    "25%": np.nanpercentile(ph, 25),
    "50%": np.nanpercentile(ph, 50),
    "75%": np.nanpercentile(ph, 75)
}
print(stats)


print("\n6 " + "=" * 72)
sulphate = data[:,9]
quality = data[:, -1]
threshold = np.nanpercentile(sulphate, 20)
mask = sulphate < threshold
avg_qlty = np.nanmean(quality[mask])

print(avg_qlty)


print("\n7 " + "=" * 72)
max_q = np.nanmax(quality)
min_q = np.nanmin(quality)

best_mask = quality == max_q
worst_mask = quality == min_q

best_mean = np.nanmean(data[best_mask], axis=0)
worst_mean = np.nanmean(data[worst_mask], axis=0)

print("best_mean: ", best_mean)
print("\nworst_mean: ", worst_mean)
```
