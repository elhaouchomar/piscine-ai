```python
import spacy

# Load the specific small model
nlp = spacy.load("en_core_web_sm")

# 1. Compute embedding for 'car' and print shape
doc = nlp("car")
vector = doc[0].vector
print(f"Shape: {vector.shape}")

# 2. Sum of first 20 values as integer ratio
first_20_sum = float(vector[:20].sum())
print(first_20_sum.as_integer_ratio())
```
