```python
import spacy

nlp = spacy.load("en_core_web_sm")

with open("data/news_amazon.txt", "r") as f:
    text = f.read()

doc = nlp(text)

for sent in doc.sents:
    for token in sent:
        if "Bezos" in token.text and token.pos_ == "PROPN":
            print(f"INFO:  {token.text} {token.pos_} {token.tag_}")
            print(f"Sentence:  {sent.text.strip()}\n")
            break
```
