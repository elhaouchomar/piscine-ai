```python
import spacy
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics.pairwise import cosine_similarity

nlp = spacy.load("en_core_web_sm")
words = "laptop computer coffee tea water liquid dog cat kitty"
doc = nlp(words)

vectors = np.array([token.vector for token in doc])

sim_matrix = cosine_similarity(vectors)

dist_matrix = 1 - sim_matrix

labels = [token.text for token in doc]
plt.figure(figsize=(10, 4))

ax = sns.heatmap(dist_matrix, annot=False, xticklabels=labels, 
                 yticklabels=labels, cmap="plasma")

ax.invert_yaxis()

plt.title("Cosine distance between words", loc='left', pad=20)
plt.savefig('w3day05ex1_plot.png')
plt.show()
```
