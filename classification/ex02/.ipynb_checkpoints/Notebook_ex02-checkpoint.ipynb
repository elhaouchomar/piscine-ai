```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-20, 15, 1000)

sigmoid = 1 / (1 + np.exp(-x))
sigmoid1 = 1 / (1 + np.exp(-(0.5 * x + 3)))
sigmoid2 = 1 / (1 + np.exp(-(5 * x + 11)))

plt.figure()
plt.plot(x, sigmoid, label="sigmoid")
plt.plot(x, sigmoid1, label="sigmoid1 (0.5x + 3)")
plt.plot(x, sigmoid2, label="sigmoid1 (0.5x + 11)")
plt.hlines(y=0.5, xmin=-20, xmax=15, color='red', linewidths=1.5)
plt.title("Sigmoids")
plt.xlabel("x")

# plt.xlim(-20, 15)
# plt.ylim(0, 1.05)
```
