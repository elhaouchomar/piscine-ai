```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

iris = load_iris()

X = pd.DataFrame(data=iris['data'], columns=iris.feature_names)
y = pd.Series(iris['target'])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=True, random_state=43
)
```

```python
def train(X_train, y_train):
    
    y0 = (y_train == 0).astype(int)
    clf0 = LogisticRegression(max_iter=1000)
    clf0.fit(X_train, y0)
  
    y1 = (y_train == 1).astype(int)
    clf1 = LogisticRegression(max_iter=1000)
    clf1.fit(X_train, y1)

    y2 = (y_train == 2).astype(int)
    clf2 = LogisticRegression(max_iter=1000)
    clf2.fit(X_train, y2)

    return clf0, clf1, clf2
```

```python
def predict_one_vs_all(X, clf0, clf1, clf2):
    
    p0 = clf0.predict_proba(X)[:, 1]  
    p1 = clf1.predict_proba(X)[:, 1]  
    p2 = clf2.predict_proba(X)[:, 1]  

    probs = np.vstack((p0, p1, p2)).T 

    classes = np.argmax(probs, axis=1)

    return classes
```

```python
clf0, clf1, clf2 = train(X_train, y_train)

y_pred_custom = predict_one_vs_all(X_test, clf0, clf1, clf2)
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

y_pred_sklearn = model.predict(X_test)

print("Custom:", y_pred_custom[:10])
print("Sklearn:", y_pred_sklearn[:10])
```
